```python
import re, collections
```

```python
text = "a sailor went to sea sea sea "+\
                  "to see what he could see see see "+\
                  "but all that he could see see see "+\
                  "was the bottom of the deep blue sea sea sea"
```

```python
def initialize_vocabulary(text):
  vocab = collections.defaultdict(int)
  words = text.strip().split()
  for word in words:
      vocab[' '.join(list(word)) + ' </w>'] += 1
  return vocab
```

```python
vocab = initialize_vocabulary(text)
print('Vocabulary: {}'.format(vocab))
print('Size of vocabulary: {}'.format(len(vocab)))
```

```python
def get_tokens_and_frequencies(vocab):
  tokens = collections.defaultdict(int)
  for word, freq in vocab.items():
      word_tokens = word.split()
      for token in word_tokens:
          tokens[token] += freq
  return tokens
```

```python
tokens = get_tokens_and_frequencies(vocab)
print('Tokens: {}'.format(tokens))
print('Number of tokens: {}'.format(len(tokens)))
```

```python
def get_pairs_and_counts(vocab):
    pairs = collections.defaultdict(int)
    for word, freq in vocab.items():
        symbols = word.split()
        for i in range(len(symbols)-1):
            pairs[symbols[i],symbols[i+1]] += freq
    return pairs
```

```python
pairs = get_pairs_and_counts(vocab)
print('Pairs: {}'.format(pairs))
print('Number of distinct pairs: {}'.format(len(pairs)))

most_frequent_pair = max(pairs, key=pairs.get)
print('Most frequent pair: {}'.format(most_frequent_pair))
```

```python
def merge_pair_in_vocabulary(pair, vocab_in):
    vocab_out = {}
    bigram = re.escape(' '.join(pair))
    p = re.compile(r'(?<!\S)' + bigram + r'(?!\S)')
    for word in vocab_in:
        word_out = p.sub(''.join(pair), word)
        vocab_out[word_out] = vocab_in[word]
    return vocab_out
```

```python
vocab = merge_pair_in_vocabulary(most_frequent_pair, vocab)
print('Vocabulary: {}'.format(vocab))
print('Size of vocabulary: {}'.format(len(vocab)))
```

```python
tokens = get_tokens_and_frequencies(vocab)
print('Tokens: {}'.format(tokens))
print('Number of tokens: {}'.format(len(tokens)))
```

```python

def tokenize(text, num_merges):
    # Initialize vocabulary
    vocab = collections.defaultdict(int)
    words = text.strip().split()
    for word in words:
        # Corrected join: ensures characters are split and </w> is at the end
        vocab[' '.join(list(word)) + ' </w>'] += 1

    for i in range(num_merges):
        # 1. Find pairs of adjacent tokens
        pairs = collections.defaultdict(int)
        for word, freq in vocab.items():
            symbols = word.split()
            for j in range(len(symbols) - 1):
                pairs[symbols[j], symbols[j+1]] += freq

        if not pairs:
            break

        # 2. Find the most frequent pair
        most_frequent_pair = max(pairs, key=pairs.get)
        print(f'Iteration {i+1}: Most frequent pair: {most_frequent_pair}')

        # 3. Create NEW vocabulary to avoid "dictionary changed size" error
        new_vocab = {}
        bigram = re.escape(' '.join(most_frequent_pair))
        p = re.compile(r'(?<!\S)' + bigram + r'(?!\S)')
        replacement = ''.join(most_frequent_pair)
        
        for word, freq in vocab.items():
            word_out = p.sub(replacement, word)
            new_vocab[word_out] = freq
        
        # Update the main vocab for the next merge iteration
        vocab = new_vocab

    # 4. Final token count calculation
    tokens = collections.defaultdict(int)
    for word, freq in vocab.items():
        word_tokens = word.split()
        for token in word_tokens:
            tokens[token] += freq

    return tokens, vocab
```

```python
tokens, vocab = tokenize(text, num_merges=40)
```

```python
print('Tokens: {}'.format(tokens))
print('Number of tokens: {}'.format(len(tokens)))
print('Vocabulary: {}'.format(vocab))
print('Size of vocabulary: {}'.format(len(vocab)))
```
